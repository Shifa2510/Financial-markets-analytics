```python
import yfinance as yf
import pandas as pd

tickers = [
    "SPY",
    "QQQ",
    "AAPL",
    "MSFT",
    "JPM",
    "GLD",
    "TLT",
    "AGG"
]

frames = []

for ticker in tickers:
    print(f"Downloading {ticker}...")
    
    data = yf.Ticker(ticker).history(
        start="2021-10-01",
        end="2026-09-30"
    ).reset_index()
    
    data["ticker"] = ticker
    
    frames.append(data)

raw = pd.concat(frames, ignore_index=True)

raw.head()
```
